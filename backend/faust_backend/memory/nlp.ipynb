```python
text="""# 桃花源后记

渔人名唤黄道远，世居武陵，打渔为生。那日从桃花源出来，一路上心里五味杂陈。

他本是个老实人，可太守听闻此事后拍案称奇，许了他好些银钱，要他带路寻找。黄道远不是没犹豫过——桃花源中人待他亲如兄弟，临别时那个白发老者还拉着他的手说：“不足为外人道也。”可太守催得紧，差役们日夜守在门口，他终究还是点了头。

然而船行三日，那条溪流竟全然变了模样。他记得清清楚楚，此处原有一块龟背巨石，溪水在此分作两股，左宽右窄，走右边那条才能到桃花源。可如今巨石还在，溪水却只有一股，直直地向前流去，哪里还有什么分岔？

再往前行，那片绚烂的桃林也不见了踪影。岸边只有寻常的芦苇和野草，秋风一吹，簌簌作响。

太守派来的差役不耐烦了，觉得这渔人是在戏耍他们。黄道远跪在船头，指天发誓说自己没有撒谎。差役们将附近的山岭翻了个遍，连个山洞的影子都没找到。最后，差役头目冷哼一声，扔下他在岸边，带着人回去了。

黄道远在溪边坐了整整一夜。月亮升起来的时候，他忽然听见水面上传来隐隐约约的笑声，像是有很多人在对歌，可仔细听时，又只剩下哗啦啦的水响。他站起身，朝着声音传来的方向喊了一声：“有人吗？”

没有人应答。

他又往前走了几步，脚下一滑，跌进了溪水里。河水并不深，可他只觉得身体往下沉，一直沉，眼前的光一点一点暗下去，像被什么柔软的东西包裹住了。

等他睁开眼，天已经亮了。他躺在河滩上，浑身湿透，手里却攥着一样东西——一片桃花瓣。深秋时节，桃花早已谢尽，这片花瓣却娇艳欲滴，仿佛刚从枝头摘下。

他将花瓣凑近鼻尖，闻到一股奇异的香气，不是花香，更像是——酒香，是那天在桃花源里喝过的米酒的味道。

从那以后，黄道远再也没有下河打过鱼。

他把渔船卖了，在溪边搭了一间茅屋，每日就坐在门口望着水面发呆。村里人都说他疯了，说他是被山里的精怪迷了心窍。他的妻子劝他，他不听；他的儿子骂他，他也不恼。只是偶尔，他会突然站起来，指着河面说：“看，桃花又开了。”

可河面上什么都没有。

日子一天天过去，黄道远老了，背也驼了，眼睛也花了。但每到春天，他总会拄着拐杖走到溪边，把一坛自家酿的米酒倒进水里。酒液顺着水流慢慢散开，他眯着眼睛看了很久，忽然笑了。

“你们喝到了吗？”他小声说，“今年的酒，比去年的甜。”

后来，黄道远在一个春天里死了。他的儿子把他埋在溪边的山岗上，面朝着那条蜿蜒的溪流。下葬那天，满山的桃花忽然全部开了，比往年早了整整半个月。花瓣纷纷扬扬地飘落下来，落在新坟上，落在溪水里，落在每一个前来送葬的人的肩头。

人们都说看见了，河道上不知什么时候多出了一条路，两岸桃花灼灼，一直延伸到很远很远的地方。路的尽头，有人撑着竹筏，笑着朝这边招手。

可等到花瓣落尽，那条路又消失了，连带着黄道远的坟，都不知怎么被桃花瓣整个掩住了，远远看去，像一座小小的、粉色的山丘。

从那以后，武陵人再也不敢轻易去找桃花源了。

只是每年春天，溪水上总会漂来几片桃花瓣，顺着水流，一路向东，像是从不曾断绝的消息。"""
```

```python
import spacy
nlp = spacy.load('zh_core_web_sm')
doc = nlp(text)
for ent in doc.ents:
    print(ent.text, ent.label_) 
```
